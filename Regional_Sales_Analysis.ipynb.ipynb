# Objective 3: Regional Sales and Profitability Analysis

This objective compares sales, profit, and profit margin across the four regions to identify the strongest and weakest regional performance.

**What this code does:** Aggregates total sales and total profit by region.

```python
region_data = df_clean.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

region_data
```

**What this code does:** Calculates regional profit margin as a percentage of sales.

```python
region_data["Profit_Margin"] = (
    region_data["Total_Profit"] / region_data["Total_Sales"]
) * 100

region_data
```

**What this code does:** Prepares a region-wise summary of sales, profit, and profit margin, sorted by sales.

```python
# REGIONAL SALES, PROFIT AND PROFIT MARGIN ANALYSIS

region_data = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
)

region_data["Profit_Margin"] = (
    region_data["Total_Profit"] /
    region_data["Total_Sales"]
) * 100

region_data.sort_values(
    "Total_Sales",
    ascending=False
).round(2)
```

**What this code does:** Plots total sales by region.

```python
# TOTAL SALES BY REGION

import matplotlib.pyplot as plt

sales_by_region = region_data.sort_values(
    "Total_Sales", ascending=False
)

plt.figure(figsize=(8, 5))

plt.bar(
    sales_by_region.index,
    sales_by_region["Total_Sales"]
)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**What this code does:** Plots total profit by region.

```python
# TOTAL PROFIT BY REGION
profit_by_region = region_data.sort_values(
    "Total_Profit", ascending=False
)

plt.figure(figsize=(8, 5))

plt.bar(
    profit_by_region.index,
    profit_by_region["Total_Profit"],
    color="green"
)

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**What this code does:** Plots profit margin by region.

```python
#PROFIT MARGIN BY REGION

margin_by_region = region_data.sort_values(
    "Profit_Margin", ascending=False
)

plt.figure(figsize=(8, 5))

plt.bar(
    margin_by_region.index,
    margin_by_region["Profit_Margin"],
    color="orange"
)

plt.title("Profit Margin by Region")
plt.xlabel("Region")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Interpretation and Results

The **West** region recorded the highest sales (**$725,457.82**) and highest profit (**$108,418.45**). The **Central** region had the lowest total profit (**$39,706.36**) among the four regions. Regional profit margins differ, so sales volume alone should not be used to judge regional performance.

# Objective 6: Future Sales Forecasting and Model Comparison

This objective compares three monthly sales forecasting methods: the 3-month moving average, the 6-month moving average, and the Seasonal Naive method. Forecast accuracy is evaluated using MAE, RMSE, and WAPE. The original forecasting comparison code and its result outputs are retained.

**What this setup code does:** Imports NumPy, which is required by the original forecasting calculation below.

```python
import numpy as np
```

**What this code does:** Builds the three forecast series, compares their accuracy using MAE, RMSE, and WAPE, selects the method with the lowest WAPE, and estimates sales for the next month.

```python
# ================================================
# OBJECTIVE 7: FORECASTING MODEL COMPARISON
# ================================================

# 1. Prepare Monthly Sales
sales = monthly_sales.copy()
sales = sales.sort_index().dropna()

# 2. Generate Three Forecasting Methods

forecast_3m = sales.shift(1).rolling(3).mean()

forecast_6m = sales.shift(1).rolling(6).mean()

seasonal_forecast = sales.shift(12)

# 3. Combine Predictions
comparison = pd.DataFrame({
    "Actual Sales": sales,
    "3-Month MA": forecast_3m,
    "6-Month MA": forecast_6m,
    "Seasonal Naive": seasonal_forecast
})

# Use the same last 12 months for fair comparison
comparison = comparison.dropna().tail(12)

# 4. Calculate Accuracy Metrics
results = []

for method in [
    "3-Month MA",
    "6-Month MA",
    "Seasonal Naive"
]:

    actual = comparison["Actual Sales"]
    predicted = comparison[method]

    mae = np.mean(np.abs(actual - predicted))

    rmse = np.sqrt(
        np.mean((actual - predicted) ** 2)
    )

    wape = (
        np.sum(np.abs(actual - predicted))
        / np.sum(np.abs(actual))
    ) * 100

    results.append({
        "Method": method,
        "MAE": mae,
        "RMSE": rmse,
        "WAPE": wape
    })

# 5. Display Comparison
results_df = pd.DataFrame(results)

print("FORECASTING METHOD COMPARISON")
print(results_df.round(2).to_string(index=False))

# 6. Select Best Method Based on WAPE
best_method = results_df.loc[
    results_df["WAPE"].idxmin(), "Method"
]

print("\nBest Forecasting Method:", best_method)

# 7. Forecast Next Month
if best_method == "3-Month MA":
    next_forecast = sales.tail(3).mean()

elif best_method == "6-Month MA":
    next_forecast = sales.tail(6).mean()

else:
    next_forecast = sales.iloc[-12]

next_month = (
    sales.index[-1] + pd.offsets.MonthBegin(1)
)

print("\nNEXT MONTH SALES FORECAST")
print("Forecast Month:", next_month.strftime("%B %Y"))
print(f"Forecasted Sales: ${next_forecast:,.2f}")

# 8. Plot Forecasting Comparison
plt.figure(figsize=(12, 6))

plt.plot(
    comparison.index,
    comparison["Actual Sales"],
    marker="o",
    label="Actual Sales"
)

for method in [
    "3-Month MA",
    "6-Month MA",
    "Seasonal Naive"
]:
    plt.plot(
        comparison.index,
        comparison[method],
        marker=".",
        linestyle="--",
        label=method
    )

plt.title("Forecasting Method Comparison")
plt.xlabel("Month")
plt.ylabel("Sales ($)")
plt.legend()
plt.xticks(rotation=30)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# 9. Final Forecast Graph
historical = sales.tail(12)

plt.figure(figsize=(12, 5))

plt.plot(
    historical.index,
    historical.values,
    marker="o",
    label="Historical Sales"
)

plt.plot(
    [sales.index[-1], next_month],
    [sales.iloc[-1], next_forecast],
    marker="o",
    linestyle="--",
    color="red",
    label="Next Month Forecast"
)

plt.title("Next Month Sales Forecast")
plt.xlabel("Month")
plt.ylabel("Sales ($)")
plt.legend()
plt.grid(alpha=0.3)
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

# Business Insights

The following five insights are based on the results of Objectives 1–6 and the supplied Sample Superstore dataset.

1. **Regional performance:** West recorded the highest sales (**$725,457.82**) and profit (**$108,418.45**). Central had the lowest total regional profit (**$39,706.36**).
2. **Sales trend:** Annual sales rose from **$164,213.48** in 2015 to **$271,489.85** in 2016, followed by a further increase to **$273,307.28** in 2017.
3. **Category and sub-category profitability:** Technology generated the highest category profit (**$145,454.95**). Tables recorded the lowest sub-category profit (**$-17,725.48**).
4. **Discount and profitability:** Discount and profit have a negative correlation of **-0.219** in the dataset. This indicates an association, not proof that discounts cause losses.
5. **Customer and forecast performance:** Consumer was the strongest customer segment by total sales and profit. In the notebook's three-method forecast comparison, Seasonal Naive had the lowest WAPE (**25.18%**) and was the selected method.

# Business Recommendations

1. **Improve Central region profitability:** Review discounting, product mix, shipping costs, and loss-making products in Central; track profit margin as well as sales.
2. **Plan around sales patterns:** Use the monthly and quarterly trends to guide inventory, staffing, and promotional timing, and investigate the drivers of the 2016 and 2017 growth.
3. **Protect product margins:** Maintain the strong Technology contribution while reviewing pricing and costs for low-profit sub-categories, especially Tables, Bookcases, and Supplies.
4. **Set discount controls:** Introduce margin-based discount limits and review high-discount transactions before approving campaigns. Measure campaign success by incremental profit, not sales alone.
5. **Use forecasts with monitoring:** Use Seasonal Naive as the best of the three tested methods, but compare each new month's actual sales with the forecast and reassess model accuracy regularly.
